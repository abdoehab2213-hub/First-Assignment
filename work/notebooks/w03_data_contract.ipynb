```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# Section 4 Code: Confirm Dataset Operational Limitations
if 'df' in locals():
    high_impressions_zero_clicks = len(df[(df['impressions_90d'] >= 1000) & (df['clicks_90d'] == 0)])
    print(f"📊 Pages with >=1000 Impressions but 0 Clicks: {high_impressions_zero_clicks:,}")
    print("💡 Limitation Verified: High impression pages with zero clicks require post-click / SERP intent metrics not present in this slice.")
else:
    print("❌ 'df' is not defined. Please run Section 1 code first.")
```
